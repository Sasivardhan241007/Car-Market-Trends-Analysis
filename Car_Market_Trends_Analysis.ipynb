11Q How many single person owned vehicles are there in this database

11ans There are 10 single owner vehicles in the database.

```python
data['Depreciation'] = data['Present_Price'] - data['Selling_Price']

most_depreciated = data.loc[data['Depreciation'].idxmax()]
least_depreciated = data.loc[data['Depreciation'].idxmin()]

print("Most Depreciated:", most_depreciated['Car_Name'], "| Value:", round(most_depreciated['Depreciation'], 2), "Lakhs")
print("Least Depreciated:", least_depreciated['Car_Name'], "| Value:", round(least_depreciated['Depreciation'], 2), "Lakhs")
```
